```python

'''
3. drinks.csv 데이터를 읽고,대륙별 술 소비량 대비 알콜 비율 컬럼(alcohol_rate_continent) 구하기
    alcohol_rate_continent = 대륙별 alcohol/대륙별전체음료소비량
   으로 계산한다.
[결과]
  country continent alcohol
0 Afghanistan AS    0.020293
1 Bahrain       AS    0.020293
2 Bangladesh  AS    0.020293
3 Bhutan        AS    0.020293
4 Brunei         AS    0.020293
'''
import pandas as pd
drinks = pd.read_csv("drinks.csv")
cols = ['country','beer','spirit','wine','alcohol','continent']
drinks.columns = cols
ncol = ['beer','spirit','wine','alcohol']

# 국가별 전체 술 소비량.
drinks["total_servings"] = drinks["beer"] +drinks["spirit"] + drinks["wine"]
continent_sum = drinks.groupby('continent').sum()

# alcohol_rate_continent 컬럼 추가
continent_sum['alcohol_rate_continent'] = continent_sum['alcohol']  / continent_sum['total_servings']

# 현재의 index를 컬럼(피처)으로 변경
continent_sum = continent_sum.reset_index()

# 대륙컬럼, 대륙별 알콜비율
continent_sum = continent_sum[['continent', 'alcohol_rate_continent']]

drinks = pd.merge(drinks, continent_sum, on='continent', how='outer')
  
print(drinks[['country', 'continent', "total_servings",'alcohol_rate_continent']].head())
#drinks[['country', 'continent', "total_servings", 'alcohol_rate_continent']].tail()
#무작위로 5개 추출
drinks[['country', 'continent', 'alcohol_rate_continent']].sample(5)
```

```python

'''
 4. 전체 평균보다 적은 알코올을 섭취하는 대륙 중에서, spirit을 가장 많이 마시는 국가 구하기
[결과]
Russian Federation
'''
import pandas as pd
drinks = pd.read_csv("drinks.csv")
cols = ['country','beer','spirit','wine','alcohol','continent']
drinks.columns = cols
ncol = ['beer','spirit','wine','alcohol']
print(drinks.loc[141])
#spirit 소비량순으로 정렬
drinks.sort_values(by=['spirit'],ascending=False)[["country","continent","spirit"]]
    
    
#전체 알콜 섭취량의 평균
total_mean = drinks.alcohol.mean()
total_mean
#대륙별 알콜 섭취량의 평균
continent_mean = drinks.groupby('continent').alcohol.mean()
continent_mean
#전체 알콜 섭취량의 평균 보다 대륙별 알콜 섭취량의 평균이 작은 대륙정보 조회
continent_under_mean = continent_mean[continent_mean < total_mean]
continent_under_mean
# 대륙 정보만 리스트로 저장.
continent_under_mean = continent_under_mean.index.tolist()
continent_under_mean

# AF,AS,OC 대륙에 속한 국가들 중 spirit_servings의 값이 
# 가장 큰 국가 조회하기
# df_continent_under_mean : 레코드들 중 대륙정보가 
#  continent_under_mean에 속한 레코드만 저장
df_continent_under_mean = drinks[drinks.continent.isin(continent_under_mean)]
print(df_continent_under_mean.sort_values(by='spirit',ascending=False))
df_continent_under_mean['spirit'].idxmax()


most_spirit_under_mean = df_continent_under_mean.loc[df_continent_under_mean['spirit'].idxmax()]
most_spirit_under_mean    
most_spirit_under_mean["country"]
```
