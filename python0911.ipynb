```python
#키값으로 value값들 조회하기
for k in member_dic.keys() :
    print("%s=>%s" % (k,member_dic[k]))

#value 값들만 조회하기
for v in member_dic.values() :
    print(v)

#key,value 값 조회하기
for i in member_dic.items() :
#    print(i) #튜플 객체
    print(i[0],i[1])

for k,v in member_dic.items() :
    print(k,v)    

for k in member_dic :  #key값들만 제공
    print("%s=>%s" % (k,member_dic[k]))
```

```python
'''
문제 : 궁합음식의 키를 입력받아 해당되는 음식을 출력하기

[예시]
음식 입력 : 라면
라면의 궁합음식은 김치입니다. 라면=>김치
음식 입력 : 불고기
등록된음식이 아닙니다.
음식 입력 :종료
프로그램을 종료합니다.
등록된음식: 
    떡볶이:오뎅
    짜장면:단무지
    라면:김치
    맥주:치킨
2. 등록된 음식이 아닌경우 
   음식을 등록하기
'''
foods ={"떡볶이":"오뎅","짜장면":"단무지","라면":"김치","맥주":"치킨"}
while True :
    myfood = input("음식 입력 : ")
    if myfood == '종료' :
         break;
    if myfood in foods :   #foods dictionary에 myfood 키값이 존재?
         print(myfood ,"=>" , foods[myfood])
    else :
         print("등록된 음식이 아닙니다")
         yn = input("등록하시겠습니까?")
         if yn.lower() == 'y' :
              f = input(myfood+"의 궁합음식:")
              foods[myfood] = f
print("프로그램을 종료합니다")
print("등록된 음식:")
for k in foods :
     print(k,":",foods[k])
```

### 컴프리헨션 (conprehension)
간단한 알고리즘을 이용해서 list,dictionary,set,tuple객체를 생성하는 방법

```python
numbers = []
#numbers list에 1 ~ 10까지의 숫자를 저장하기
for n in range(1,11) :
    numbers.append(n)
print(numbers)  

numbers = [x for x in range(1,11)]
print(numbers)

```

```python
#2 부터 20까지의 짝수들을 numbers 리스트에 저장하기
numbers = [x for x in range(2,21,2)]
print(numbers)

numbers = [x*2 for x in range(1,11)]
print(numbers)

numbers = [x for x in range(1,21) if x % 2 == 0]
print(numbers)
```

```python
# 1 ~ 10까지의 수 중 2또는 3의 배수만 가진 nums 리스트 생성하기
nums = [n for n in range(1,11) if (n%2==0) or (n%3==0)]
print(nums) # [2,3,4,6,8,9,10]
```

```python
#두개의 리스트의 값을 하나씩 매칭하여 튜플로 생성하기
clist = ["black","white"]
slist = ['S','M','L']
dlist = [(c,s) for c in clist for s in slist]
print(dlist)
```

```python
products = {'냉장고':220,'건조기':140,'TV':130,'세탁기':150,'오디오':50,'컴퓨터':250}
# 200만원 미만의 제품만 product1 dictionary에 저장하기
product1 = {}
for k in products :
    if products[k] < 200 :
        product1[k] = products[k]
print(product1)

product1 = {}
for k,v in products.items() :
    if v < 200 :
        product1[k] = v
print(product1)

#컴프리헨션 방식
product1 = {k:v for k,v in products.items() if v < 200}
print(product1)
```

```python
#dictionary 데이터 정렬하기
import operator
dic1,list1 = {}, []
dic1 = {"Thomas":"토마스","Edward":"에드워드","Henry":"헨리","Gothen":"고든","James":"제임스"}
print(dic1.items())
''' 
sorted : dic1.items()의 데이터를 key속성의 값 기준으로 정렬함
   operator.itemgetter(0) :0번인덱스의 데이터 기준. 영문기준
   reverse=True : 내림차순 정렬
   => dic1 데이터의 영문의 내림차순 정렬
'''
list1 =sorted(dic1.items(),key=operator.itemgetter(0),reverse=True)
print(list1)
#영어이름 순으로 정렬 출력하기
list1 =sorted(dic1.items(),key=operator.itemgetter(0),reverse=False)
print(list1)
#한글이름 순으로 정렬 출력하기
list1 =sorted(dic1.items(),key=operator.itemgetter(1))
print(list1)
```

```python
#매개변수
def add(v1,v2) :
    return v1 + v2

def subs(v1,v2) :
    return v1 - v2
```

```python
hap,sub = 0,0
hap = add(10,20) #함수의 호출. 호출시 인자값과 매개변수의 갯수가 맞아야 함.
print("10 + 20 =",hap)
sub = subs(10,20)
print("10 - 20 =",sub)
```

```python
# 한개의 함수에서 여러개의 리턴값을 전달. 리스트로 전달함
def add_sub(v1,v2) :
    list1 = []
    list1.append(add(v1,v2))
    list1.append(subs(v1,v2))
    return list1
```

```python
list1 = add_sub(10,20)
print(list1[0],list1[1])
```

```python
#가변 매개변수 ; 매개변수의 갯수 지정을 안해도 됨
def valparam(* p) : #0개이상의 매개변수가 가능
    result = 0
    for i in p :
        result += i
    return result
```

```python
print("valparam()=",valparam())
print("valparam(10)=",valparam(10))
print("valparam(10,20)=",valparam(10,20))
print("valparam(10,20,30)=",valparam(10,20,30))
```

```python
# 숫자형 리스트를 매개변수로 하는 함수 getMean과 getSum 함수를 구현하기
# getMean 함수 기능: 리스트 요소의 평균 리턴
# getSum 함수 기능: 리스트 요소의 합계 리턴
def getSum(l) :
    return sum(l)
def getMean(l) :
    #return sum(l) / len(l)
    return sum(l) / len(l) if len(l) > 0 else 0. #조건 연산자
```

```python

list1=[2,3,3,4,4,5,5,6,6,8,8]
print("list1 의 합:",getSum(list1))
print("list1 의 평균:",getMean(list1))

list2=[2,6,8,8]
print("list2 의 합:",getSum(list2))
print("list2 의 평균:",getMean(list2))

list3=[]
print("list3 의 합:",getSum(list3))
print("list3 의 평균:",getMean(list3))
```

### 람다식

```python
#람다방식으로 함수 변경
hap2 = lambda num1,num2 : num1 + num2
print(hap2(10,20))
print(hap2(100,200))
```
