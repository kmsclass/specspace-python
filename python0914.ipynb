### 클래스 : 객체지향언어. 다중상속이 가능함
사용자 정의 자료형  
구조체(멤버변수, 필드) + 함수(멤버메서드,멤버함수)  
기본생성자 : 클래스내에 생성자가 없으면 시스템이 자동으로 생성해 주는 생성자  
          형태 : __init__(self). 
self : 자기 참조 변수

```python
class Car :  #기본생성자 제공 클래스.
    color = "" #필드
    speed = 0  #필드
    def upSpeed(self,value) :  #멤버함수
        self.speed += value
    def downSpeed(self, value) : #멤버함수
        self.speed -= value
```

```python
#  객체화
mycar1 = Car()   #Car 클래스의 생성자(__init__) 호출
mycar1.color = "빨강"
mycar1.speed = 30
mycar2 = Car()   #Car 클래스의 생성자 호출
mycar2.color = "검정"
mycar2.speed = 50
print("자동차1의 색상",mycar1.color,"속도는",mycar1.speed,"입니다")
print("자동차2의 색상",mycar2.color,"속도는",mycar2.speed,"입니다")
```

```python
# 생성자 구현하기
class Car :  
    color = "" #필드
    speed = 0  #필드
    def __init__(self,c,s) : # 생성자 구현. 기본생성자없음
        self.color = c
        self.speed = s
    def upSpeed(self,value) :  #멤버함수
        self.speed += value
    def downSpeed(self, value) : #멤버함수
        self.speed -= value
```

```python
mycar1 = Car("빨강",30) 
mycar2 = Car("검정",50) 
print("자동차1의 색상",mycar1.color,"속도는",mycar1.speed,"입니다")
print("자동차2의 색상",mycar2.color,"속도는",mycar2.speed,"입니다")
```

```python
#클래스에서 사용되는 특별한 함수들
class Line :
    length = 0
    def __init__(self,length) :
        self.length = length
    #def __repr__(self) :  #객체의 문자열화
    #    return "선의길이:" + str(self.length)
    def __str__(self) :  #객체의 문자열화
        return "선의길이:" + str(self.length)
    def __add__(self,other) :
        print("+연산자 호출됨",end="")
        return self.length + other.length
    def __lt__(self,other) :    
        print("< 연산자 호출:",end="")
        return self.length < other.length
    def __eq__(self,other) :   
        print("== 연산자 호출:",end="")
        return self.length == other.length
    def __gt__(self,other) :   
        print("> 연산자 호출:",end="")
        return self.length > other.length
```

```python
#Line 클래스의 객체를 2개 생성하기. length 의 값이 200, 100인 객체로 생성하기
line1 = Line(200)
line2 = Line(100)
print(line1) #__repr__ 함수 호출
print(line2)
print("두선의 길이의 합:",line1+line2) #__add__ 함수호출
print("두선의 길이의 합:",line1.__add__(line2))
if line1 < line2 :   # __lt__
    print("line2 선이 더 깁니다.")
elif line1 == line2 :   #__eq__
    print("line1과 list2 선의 길이는 같습니다.")
elif line1 > line2 :    #__gt__
    print("line1 선이 더 깁니다.")
```

```python
#주민번호 감추기 : data 문자열에서 주민번호형식인 경우 뒤 7자리를 *로 치환하기
data = '''     
   park 800905-1234567
   kim  700905-2234567
   choi 850101-a123456
'''
print(data)
#[park 800905-*******,kim 700905-*******,choi 850101-a123456]
result = [] #리스트. 결과 저장

for line in data.split("\n"): #[park 800905-1234567,kim  700905-2234567,choi 850101-a123456]
    word_result = [] #[park,800905-*******]
    for word in line.split(" ") : # line.split(' ') : [park, 800905-1234567]
        if len(word) == 14 and word[:6].isdigit() and word[7:].isdigit() :
                word = word[:6]+"-"+"*******" 
        word_result.append(word)     
    result.append(" ".join(word_result))   #word_result리스트의 요소들을 " "으로 연결하여 문자열로 생성 
print("\n".join(result)) 
```

```python
#정규화 모듈로 구현하기
import re #정규화 사용하기 위한 모듈
pat = re.compile(r"(\d{6,7})[-]\d{7}")  #패턴 설정 : 숫자6~7 - 숫자7자리
print(pat.sub(r"\g<1>-*******",data)) #data 중 패턴에 맞는 문자열을 변경하기 
'''
(\d{6,7})[-]\d{7}

( ) : 그룹
\d : 숫자
{6,7} : 자리수 6~7자리 
\d{6,7} : 6~7자인인 숫자
[-] : - 문자
\d{7} : 숫자 7자리
g<1> : 첫번째 그룹.\d{6,7}

pat.sub(변경될형식, 원본데이타) : pattern에 맞는 문자열을 치환
'''
```

```python
str1 = "The quick brown fox jumps over the laze dog Te The Thhhhe"
str_list = str1.split() #공백으로 문자열 분리
print(str_list)
#Th*e : T로 시작하고,  h문자의 갯수가 0개이상이고, e문자로 종료하는 문자열
pattern = re.compile("Th*e") 
count = 0
for word in str_list :
    if pattern.search(word) :  #패턴과 일치?
        print(word,end=",") 
        count += 1
print()
print("결과1 : {0:s}:{1:d}".format("갯수",count))

#대소문자 구분없이 검색
pattern = re.compile("Th*e",re.I) #re.I:대소문자 구분없이
count = 0
for word in str_list :
    if pattern.search(word) :  #패턴과 일치?
        print(word,end=",") 
        count += 1
print()
print("결과2 : {0:s}:{1:d}".format("갯수",count))

#str1문자열에서 pattern에 맞는 문자들을 리스트로 출력
print("결과 3 :",re.findall(pattern,str1))
print("결과 4 일치하는 문자열의 갯수:",len(re.findall(pattern,str1)))
print("결과 5 일치하는 문자열 치환:",pattern.sub('aaa',str1))
```

```python
# sqlite : 파이썬 내부에 존재하는 DBMS
# 데이터 베이스에 table 생성하기
import sqlite3
dbpath="test.sqlite" #db 저장할 공간의 이름지정
# conn : sqlite 파일과 연결된 객체
conn = sqlite3.connect(dbpath) #db저장 공간 설정
cur = conn.cursor() #sql문장을 실행을 위한 객체
#executescript : 여러개의 문장들 한꺼번에 실행할 수 있는 함수
#                한문장의 끝에 ;으로 구분된 문장들을 실행
cur.executescript('''
  drop table if exists items;
  create table items (item_id integer primary key,
  name text unique, price integer);
  insert into items (name,price) values ("Apple",800);
  insert into items (name,price) values ("Orange",500);
  insert into items (name,price) values ("Banana",600);
''')
conn.commit() #트랜젝션 완료. 물리적 공간에 완성
```
