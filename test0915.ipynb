```python

'''
1. 데이터 베이스의 테이블을 조회하기
select 구문을 입력하고 다음과 같은 결과가 출력되도록 프로그램을 작성하시오
[결과]
sql 입력하세요=========
select * from items

조회 레코드수: 3 ,조회 컬럼수: 3
(1, 'Apple', 800)
(2, 'Orange', 500)
(3, 'Banana', 600)
'''
import sqlite3
from sqlite3 import OperationalError
conn = sqlite3.connect("test.sqlite")
cur = conn.cursor()
while True :
    sql = input("sql 입력하세요=========\n")
    if sql=="" :
        break
    try :
        cur.execute(sql)
    except OperationalError as e :
        print("sql 문장에 오류가 있습니다.", end="=>")
        print(e)
    else :    
        rows = cur.fetchall()
        print("조회 레코드수:",len(rows),",조회 컬럼수:",len(rows[0]))    
        for row in rows :
            print(row)
cur.close()
conn.close();
```

```python
"""
5. 화면에서 서로다른 4자리 숫자를 입력받아 야구 게임하기
   컴퓨터가 서로다른 4자리 숫자를 임의로 기억하고,
   사용자는 숫자를 입력하여 컴퓨터가 저장한 데이터를 맞춘다.
   입력한 숫자와, 컴퓨터가 저장한 숫자가 일치하는 경우
   자리수가 맞으면 strike,
   자리수가 틀리면 ball
   strike과 ball의 갯수를 알려준다
[결과]
서로다른 4자리 숫자를 입력하세요: 1234
Strike: 0 Ball: 2

서로다른 4자리 숫자를 입력하세요: 5678
Strike: 0 Ball: 2

서로다른 4자리 숫자를 입력하세요: 8146
3 번만에 맞췄습니다.
"""

import random 
 
list1=[]
set1 = set(list1) #중복되지 않도록 데이터를 저장
while len(set1) < 4 :
    rnum = random.randrange(0,10) #0~9사이의 값
    set1.add(rnum)
list1 = list(set1) #컴퓨터가 저장하고 있는 숫자
print(list1) 
cnt = 0
while True :
    number = input("서로다른 4자리 숫자를 입력하세요: ")
    cnt += 1
    strike = 0
    ball = 0
    for n in number:
        num = int(n) 
        if list1.count(num)  == 1:
            if number.find(n) == list1.index(num):
                strike += 1
            else:
                ball += 1
                
    if(strike == 4) :
        break                
    else :
        print("Strike:", strike, "Ball:", ball)

print(cnt,"번만에 맞췄습니다.")   
```

```python
'''
6. 자연수를 입력받아 입력받은 수 N이 짝수이면 N!을,
   홀수이면 ΣN을 구하는 함수
   calc 함수를 작성하기
   4! = 4 * 3 *2 * 1
   Σ5 = 5 + 4 + 3+ 2 + 1
'''
def calc(n):
    if n % 2 == 0:
        print(str(n)+"!",end="=>")
        result = 1
        for i in range(1, n + 1):
            result *=  i
    else:
        print("Σ" + str(n),end="=>")
        result = 0
        for i in range(1, n + 1):
            result +=  i
    return result

def calc1(n):
    if n % 2 == 0:
        print(str(n)+"!",end="=>")
        fac = lambda x :  x if x==1 else x*fac(x-1)
        return fac(n) #4
    else:
        print("Σ" + str(n),end="=>")
        hap = [x for x in range(n+1)]
        return sum(hap)

num = int(input("숫자를 입력하세요")) 
print(calc(num))
```
