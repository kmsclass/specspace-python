### for 구문
range(1 , 101) : 1부터 101 앞까지 , 100까지의 숫자
range(1 , 101,2) : 1부터 101 앞까지 , 100까지의 숫자. 1부터 2씩 증가. 1,3,5,7
i : 반복문 내에서 변경되는 값

```python
sum = 0
for i in range(1 , 101) :
    sum += i #sum : 1+2+3...+100
print ("1 ~ 100까지의 합:",sum)    
```

### while 구문
조건의 결과가 참인 동안 반복문 수행

```python
num = 1
sum = 0
while num <= 100 :
    sum += num
    num +=1
print ("1 ~ 100까지의 합:",sum) 
```

컴퓨터가 1 ~ 99사이의 임의의 수를 저장하고,
사용자는 숫자를 입력받아서 컴퓨터가 저장한 숫자를 맞추기
컴퓨터의 숫자와 입력된수를 비교하여 큰수, 작은수인지 출력
정답 입력시 입력한 횟수를 출력하기
1. 임의의 수 : 난수
2. 정답을 맞추는 동안 계속 입력받기. => 반복문
3. break 구문 

```python
#1. 난수 모듈
import random
rnum = random.randrange(1,100) #1~99 사이의 임의의 수 리턴
cnt = 0
while True : # True. 조건문의 결과가 참인동안 반복수행
    a = int(input("1 ~ 99 사이의 숫자를 입력하세요: "))
    cnt += 1
    if a > rnum :
        print("작은수")
    elif a < rnum :
        print("큰수")
    else :
        print("정답입니다.")
        break;  #반복문 종료
print("%d번 만에 정답을 맞췄습니다." % cnt)    
```

문제  
삼각형의 높이를 입력받아 삼각형을 *로 출력하기  
삼각형의 높이 : 3  
*

```python
h = int(input("삼각형의 높이 :"))
for i in range(1,h+1) :
    print("*"*i)
print()

for i in range(h,0,-1) :
    print("*"*i)
print()

for i in range(1,h+1) :
    print(" " * (h - i), end="")
    print("*"*i)
print()

for i in range(h,0,-1) :
    print(" " * (h - i), end="")
    print("*"*i)
print()
```

```python
#중첩반복문 : 반복문 내에 반복문이 존재
i,j = 0,0  #변수 초기화
for i in range(2,10) : #2 ~ 9단,  i : 3
    print("%5d단" % i) 
    for j in range(2,10) : # j : 2
        print("%2d X %2d = %3d" % (i,j,(i*j))) # 2 X 9 = 18
    print() 
```

```python
# 구구단을 가로로 출력
for i in range(2,10) :
    #%4d : %d:10진정수 %4d: 4자리로 출력
    #%4s : %s:문자열출력
    #end="" : 출력되는 마지막 컬럼. 기본값:\n
    print("%4d단%4s" % (i," "),end="") 
print()

for j in range(2,10) : #j:3
    for i in range(2,10) : #i:4
        print("%2dX%2d=%3d " % (i,j,(i*j)),end="")
    print() 
```
