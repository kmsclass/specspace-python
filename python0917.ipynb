```python
import numpy as np
import csv
import matplotlib.pyplot as plt
```

```python
f = open("age.csv","r",encoding="CP949")
data = csv.reader(f) #f 파일을 csv 형식의 데이터로 읽기
print(type(data))
next(data) #한줄 skip
name = "역삼1동"
for row in data :
    if row[0].find(name) >= 0 :
        print(name,"=",row[0])
        row = list(map((lambda x : x.replace(",","")),row))
        print(row)
        home = np.array(row[3:],dtype=int) #list 데이터를 배열로 변경. 0세부터 데이터부분을 정수형 배열
        print(home)
        break

plt.style.use('ggplot')
plt.rc('font',family='AppleGothic')
plt.figure(figsize=(10,5),dpi=100)
plt.title(name + " 지역의 인구 구조")
plt.plot(home)
```

```python
# age.csv 파일을 이용해서 선택한 지역의 인구구조와, 
# 가장비슷한 인구구조를 가진 지역의 그래프와 지역을 출력하기
# 선택한지역은 한개만 가능하도록 한다.
import re #정규식 모듈
f=open("age.csv","r",encoding="CP949")
data=csv.reader(f)
next(data) 
data=list(data) #csv객체 => 리스트 객체
name='신림동' 
mn=1
result_name = ''
result=0
for row in data :
    if name in row[0] : #row[0]에 name 존재?
       home_name = re.sub('\(\d*\)', '', row[0])
       row = list(map((lambda x:x.replace(",","")),row))
       #row[2] : 총인구수
       # 나이인구수/총인구수 => 비율
       home = np.array(row[3:], dtype =int) /int(row[2]) 

for row in data :
    row = list(map((lambda x:x.replace(",","")),row))
    #away : 비교 대상이 되는 지역의 데이터
    away = np.array(row[3:], dtype =int) /int(row[2]) 
    s = np.sum((home - away) **2)  #mse값 구함
    if s < mn and name not in row[0] :  #mse가 작은 지역 선택. home은 아님
        mn = s
        result_name = re.sub('\(\d*\)', '', row[0])  #지역명
        result = away         #mse가 작은 지역의 데이터

plt.figure(figsize = (10,5), dpi=100)            
plt.title(name +' 지역과 가장 비슷한 인구 구조를 가진 지역')
plt.plot(home, label = home_name)
plt.plot(result, label = result_name)
plt.legend()
```

```python
# 판다스를 이용하여 구현하기
import pandas as pd
#index_col=0 : 첫번째 컬럼을 인덱스로 설정
#thousands=',' : 숫자에 천단위로 , 존재. 숫자로 인식할때 , 감안
df = pd.read_csv('age.csv', encoding='cp949',index_col=0,thousands=',')
#컬럼명 설정
col_name=['총인구수','연령구간인구수']
for i in range(0,101) :  #0 ~ 100 숫자
    col_name.append(str(i)+'세')
df.columns=col_name   #현재 컬럼의 이름을 변경
#df.div : 행의 모든 값을 총인구수로 나누기 
df = df.div(df['총인구수'],axis=0) 
#컬럼 제거하기
del df['총인구수'],df['연령구간인구수']
#모든 데이터들은 비율변경됨
name='신림동'
df.fillna(0, inplace=True)  #결측값이 존재하면 0으로 변경
# df.index : 행정구역 이름. 
a=df.index.str.contains(name) #기준 동명을 가진 인덱스를 검색
df2 = df[a]  #기준 동명의 데이터. 신림동 데이터
b = list(map((lambda x : not x),a))  # 신림동이 아닌 데이터가  True
df3 = df[b] #기준 동명이 아닌 데이터. 신림동이 아닌 다른 데이터

mn = 1
#df3.T : 0세,1세 ... 100세 컬럼을 행으로 변경
for label, content in df3.T.items():  #컬럼값,데이터
    s = sum((content - df2.iloc[0])**2)  #mse 값
    if s < mn :
        mn = s
        result = content   #기준동과 인구구조가 가장 가까운 데이터
        result_name =  label #인구구조가 가장 가까운 동의 이름
df2.T.plot() 
result.plot(legend=result_name, figsize=(15,5))
plt.legend()
```

```python
#new_Date 컬럼에서 년,월,일 정보를 각각의 컬럼으로 저장하기
df['Year'] = df['new_Date'].dt.year
df['Month'] = df['new_Date'].dt.month
df['Day'] = df['new_Date'].dt.day
df.info()
```

```python
df.select_dtypes(exclude=['string','datetime'])
```

```python
#월별 평균값 조회하기
df.select_dtypes(exclude=['string','datetime']).groupby("Month").mean()
```
