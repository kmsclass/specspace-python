### sql
DDL : 테이블을 생성, 변경, 제거  
    create : 테이블 생성 명령어
    alter  : 테이블 구조 변경 명령어
    drop   : 테이블 제거 명령어
DML : 테이블내에 저장된 데이터를 추가, 변경, 삭제, 조회 
    select(R) : 데이터 조회 명령어
    insert(C) : 데이터 추가 명령어
    update(U) : 데이터 변경 명령어
    delete(D) : 데이터 삭제 명령어

    CRUD : 데이터를 추가,조회,수정,삭제 용어

```python
import sqlite3
dbpath="test.sqlite" #db 저장할 공간의 이름지정
# conn : sqlite 파일과 연결된 객체
conn = sqlite3.connect(dbpath) #db저장 공간 설정
cur = conn.cursor()
cur.execute("select * from items")
item_list = cur.fetchall() #실행된 결과를 리스트로 리턴
print(item_list)
print(type(item_list))

for item_id, name,price in item_list :
    print(item_id, name, price)
```

```python
# mydb sqlite db 생성하기
# mydb db에 usertable 테이블 생성하기
# id char(4) primary key, username char(15), email char(50),
#  birthyear int 컬럼을 가진 테이블 생성하기

import sqlite3
conn = sqlite3.connect("mydb")
cur = conn.cursor()
cur.executescript(
    '''
drop table if exists usertable;
create table usertable (
   id char(4) primary key,
   username char(15),
   email char(50),
   birthyear int
 );
    '''
)
```

```python
#  화면에서 id, 이름, 이메일, 출생년도 입력받아 db에 등록하기
while True :
    d1 = input("사용자ID : ")
    if d1 == '' :
       break
    d2 = input("사용자이름 : ")
    d3 = input("이메일 : ")
    d4 = input("출생년도 : ")
    sql = "insert into usertable (id, username,email, birthyear) values ('" \
    +d1 + "','" + d2 +"','" + d3 + "'," + d4+")"
    print(sql)
    cur.execute(sql)
    conn.commit()
conn.close()
```

```python
#여러개의 레코드 추가하기
import sqlite3
data = [('t5','테스트5','test5@aaa.bbb',1992),
        ('t6','테스트6','test6@aaa.bbb',1993),
        ('t7','테스트7','test7@aaa.bbb',1993)]
sql = "insert into usertable (id, username,email, birthyear) values (?,?,?,?)"
conn = sqlite3.connect("mydb")
cur = conn.cursor()
cur.executemany(sql,data) #여러건의 레코드 추가
conn.commit()
```

```python
'''
score.txt 파일의 내용
 홍길동,100
 김삿갓,50
 이몽룡,90
 임꺽정,80
 
 문제:score.txt 파일을 읽어서 점수의 총점과 평균 구하기
'''
import re
infp = open("score.txt","r",encoding="UTF8")
instr = infp.read()
pattern = re.compile(r"\d+")
list1 = re.findall(pattern,instr)
list1 = list(map(int, list1))
print("총점:",sum(list1),",  평균:", sum(list1)/len(list1))
```

```python
#2진파일을 읽어서 이진파일로 저장하기
#apple.gif 파일을 읽어서 apple2.gif 파일로 저장하기
#apple.gif 파일 읽기
infp = open("apple.gif","rb")
#apple2.gif 파일에 쓰기
outfp = open("apple2.gif","wb")
while True :
    indata = infp.read(8096) #8096바이트만큼 읽기
    if not indata :
        break
    outfp.write(indata)
infp.close()
outfp.flush()
outfp.close()
```

```python
#with 구문 : 자동으로 close를 해주는 기능
#csv파일 읽기 : 텍스트파일. 단어간의 ,로 구분됨
#jeju1.csv 파일을 jeju2.csv로 저장하기
# 맥 : 기본 인코딩 UTF8
# 윈도우 : 기본 인코딩 CP949
with open('jeju1.csv','r',encoding="UTF8") as infp :
    with open("jeju2.csv","w",encoding="utf8") as outfp :
        header = infp.readline()
        print(header)
        header = header.strip()
        header_list = header.split(",") #["장소","LON","LAT"]
        outfp.write(",".join(map(str,header_list)) + "\r\n")
        for row in infp :  #infp 파일에서 한줄씩 데이터 읽기
            outfp.write(row)
```
