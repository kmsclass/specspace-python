```python
# 한줄 주석 : 이후 부분이 실행안됨. 파이썬 문법을 따르지 않아도 됨. 
'''
여러줄 주석 : 문자열 인식함. 여러줄 문자열임
'''
print('안녕하세요')
print("""동해물과 백두산이
마르고 닳도록
하느님이 보우하사""")
print('''동해물과 백두산이
마르고 닳도록
하느님이 보우하사''')
```

```python
#문자열의 인덱스 사용가능함
print('안녕하세요'[0]) #0번인덱스의 문자 출력
print('안녕하세요'[0:2]) #0번인덱스 부터 2번인덱스 앞까지의 문자 출력
print('안녕하세요'[:2]) #0번인덱스 부터 2번인덱스 앞까지의 문자 출력
print('안녕하세요'[2:5]) #2번인덱스 부터 4번인덱스 까지의 문자 출력
print('안녕하세요'[2:]) #2번인덱스 부터 끝까지의 문자 출력
print('안녕하세요'[::2]) #2번인덱스 부터 끝까지의 문자 출력
print('안녕하세요'[-1]) #끝 문자 출력
print('안녕하세요'[::-1]) #역순으로 출력
```

```python
#len : 문자열의 길이
print(len("안녕하세요"))
```
